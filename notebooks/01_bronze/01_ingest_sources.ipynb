```python
crm_customers = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(crm_path + "cust_info.csv")
)

display(crm_customers)
```

```python
crm_customers.printSchema()
```

```python
print("Number of rows:", crm_customers.count())
```


## CRM Products 

```python
crm_products = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(crm_path + "prd_info.csv")
)

display(crm_products)
```
